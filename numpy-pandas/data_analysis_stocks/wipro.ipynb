```python
df['Date'] = pd.to_datetime(df['Date'])
df.dtypes
```

```python
df['year'] = df['Date'].dt.year
```

Line graph - year vs mean close price

```python
temp=df.groupby('year')['Close'].mean()
plt.plot(temp.index, temp)
plt.title('Year vs Average close price of Wipro stock')
plt.xlabel('Years')
plt.ylabel('Average close prices')
plt.show()
```

```python
temp=df.groupby('year')['Turnover'].mean()
plt.plot(temp.index, temp)
plt.title('Year vs Average Turn over of Wipro stock')
plt.xlabel('Years')
plt.ylabel('Average close prices')
plt.show()
```

```python
temp=df.groupby('year')['Volume'].mean()
plt.plot(temp.index, temp)
plt.title('Year vs Average Volume of Wipro stock')
plt.xlabel('Years')
plt.ylabel('Average close prices')
plt.show()
```

```python
df['month'] = df['Date'].dt.month

temp=df[df['year']==2020]
temp1=temp.groupby('month')['Close'].mean()
plt.plot(temp1.index, temp1)
plt.title('Average close price by moth in 2020')
plt.xlabel('Months')
plt.ylabel('Average close prices in months')
plt.show()
```

```python
def yearWiseGrapghPlot(year):
    df['month'] = df['Date'].dt.month
    temp=df[df['year']==year]
    temp1=temp.groupby('month')['Close'].mean()
    if len(temp1) > 0:
        plt.plot(temp1.index, temp1)
        plt.title(f'Average close price by months of year : {year}')
        plt.xlabel('Months')
        months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
        plt.xticks(range(1, 13), months)
        plt.ylabel('Average close prices in months')
        plt.show()
for i in range(1920, 2030):
    yearWiseGrapghPlot(i)
```

```python
def yearWiseGrapghPlot(year):
    df['month'] = df['Date'].dt.month
    temp=df[df['year']==year]
    temp1=temp.groupby('month')['Close'].std()
    if len(temp1) > 0:
        plt.plot(temp1.index, temp1)
        plt.title(f'Average close price by months of year : {year}')
        plt.xlabel('Months')
        months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
        plt.xticks(range(1, 13), months)
        plt.ylabel('Average close prices in months')
        plt.show()
for i in range(1920, 2030):
    yearWiseGrapghPlot(i)
```

```python
df[df['year']!=2000].groupby('month')['Close'].mean().plot(kind='line', label='mean')
df[df['year']!=2000].groupby('month')['Close'].median().plot(kind='line', label='median')
df[df['year']!=2000].groupby('month')['Close'].min().plot(kind='line', label='min')
df[df['year']!=2000].groupby('month')['Close'].max().plot(kind='line', label='max')
plt.legend()
plt.show()
```
