Pandas

Pandas is a powerful Python library for data manipulation and analysis.
It provides data structures like Series (1D) and DataFrame (2D) to handle 
and process data efficiently. Below is an explanation of how to work with 
1D and 2D data in Pandas:

```python
# Panel Data
# data analysis and manipulation
# series (1D) - rows/ columns singularlypip 
# data frame (2D) - table format
```

Series (1D)

```python
import pandas as pd

# Create a Series from a list
data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print(series)

series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])
print(series)
```

DataFrame (2D)

```python
# Create a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
df = pd.DataFrame(data)
print(df)

# Access a single column (returns a Series)
print('\n=================\n', df['Name'])

# Access multiple columns
print('\n=================\n', df[['Name', 'City']])

# Access a row by index
print('\n=================\n', df.iloc[1])  # Second row (index 1)
```

List of list

```python
df=pd.DataFrame([[1, 2, 3],["No Data", 22, 23]], columns=['age', 'mark1', 'mark2'])
print(df)
print("\n=======================\n")
# columns must be homogeneous, other wise it will be considered as an object
print(df.dtypes)
```

2D array to Dataframe

```python
# 2D array of shape (10, 5) with random numbers b/w 1-50 and convert into dataframe
import numpy as np
array1 = np.random.randint(1,51,(10,5))
print(array1)
print("\n=======================\n")
df3=pd.DataFrame(array1, columns=['a', 'b', 'Column 3', 'c', 'd'])
print(df3)
print("\n=======================\n")
```

.loc (Location) & .iloc (Integer Location)

```python
# Exercise: DataFrame Operations in Pandas
# Step 1: Create a DataFrame
# Create a DataFrame with the following data:
data = {"ID": [1, 2, 3, 4, 5],
        "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
        "Age": [25, 30, 35, 40, 28],
        "City": ["New York", "Chicago", "Boston", "Seattle", "Austin"],
        "Salary": [50000, 60000, 55000, 70000, 65000]}
df = pd.DataFrame(data)
print('\n=================\n', df,'\n=================\n')
```

```python
# Step 2: Locate Specific Rows and Columns
# 1. Select and print the row where Name is "Charlie".
# 2. Retrieve the Name and Salary columns for all employees.
print('\n=================\n', df.iloc[2, :],'\n=================\n')
print('\n=================\n', df["Name", "Salary"][df['Name']=='Charlie'],'\n=================\n')
print('\n=================\n', df.loc[:,["Name", "Salary"]],'\n=================\n')
```

```python
# Step 3: Drop Columns and Rows
# 1. Drop the City column without modifying the original DataFrame.
# 2. Drop the row where ID is 3 permanently.
df.drop(columns=['City'])
df.drop(index=2, inplace=True)
print(df)
```

```python
# Step 4: Add a New Column
# 1. Add a new column Experience with values [2, 5, 8, 10, 4].
df['Experience'] = [2, 5, 8, 10, 4]
print('\n=================\n', df,'\n=================\n')
```

```python
# Step 5: Understand the inplace Parameter
# 1. Drop the Age column using inplace=True.
# 2. Try accessing the Age column again.
if ('Age' in df.columns):
    df.drop(columns='Age', inplace=True)
df['Age']
```

```python
# Step 6: Reset Index
# 1. Reset the index after dropping the row and store it back in the same DataFrame.
df.reset_index(drop=True, inplace=True)
print('\n=================\n', df,'\n=================\n')
```

```python
# Step 7: Call Multiple Columns
# 1. Select and print only Name, Salary, and Experience.
print('\n=================\n', df[['Name', 'Salary', 'Experience']],'\n=================\n')

# Step 8: Rename Columns
# 1. Rename Salary to Annual_Income and Experience to Years_of_Experience.
df.rename(columns={"Salary":"Annual_Income"}, inplace=True)
print('\n=================\n', df,'\n=================\n')
```
