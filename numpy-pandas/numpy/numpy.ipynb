```python
# ---
# Exercise 1: Boolean Indexing on a 1D Array
# Given the following array:
# import numpy as np  
# arr = np.array([12, 45, 7, 19, 34, 28, 90, 55, 63])
# Tasks:
# 1. Extract all elements greater than 30.
# 2. Extract all odd numbers.
# 3. Replace all numbers that are multiples of 5 with -1.
# 4. Count how many numbers are less than 50.
arr1 = np.array([12, 45, 7, 19, 34, 28, 90, 55, 63])
print(arr1[arr1>30])
print(arr1[arr1%2!=0])
arr1[arr1%5==0]=-1
print(arr1)
print(arr1[arr1>50])
print(len(arr1[arr1>50]))
```

```python
# ---
# Exercise 2: Boolean Indexing on a 2D Array
# Consider the following 2D array:
# arr = np.array([[10, 25, 40],  
#                 [55, 18, 72],  
#                 [91, 30, 64]])
# Tasks:
# 1. Extract all numbers greater than 50.
# 2. Replace all even numbers with 0.
# 3. Find the sum of all numbers that are divisible by 5.
# 4. Count how many numbers are greater than 20 but less than 70.
arr = np.array([[10, 25, 40],  
                 [55, 18, 72],  
                 [91, 30, 64]])
arr[((arr>20) & (arr<70))]
```

```python

# ---
# Exercise 3: Fancy Indexing
# Given the following array:
# arr = np.array([[12, 34, 56],  
#                 [78, 90, 23],  
#                 [45, 67, 89]])
# Tasks:
# 1. Extract elements from the first and third row.
# 2. Extract the second column from the array.
# 3. Extract specific elements: 12, 90, 67 using advanced indexing.
# 4. Swap the first and last row of the array.
print(arr[[0,2]])
arr = np.array([[12, 34, 56],  
                [78, 90, 23],  
                [45, 67, 89]])
arr[[0,2]]=arr[[2,0]]
print(arr)
```

```python
# ---
# Exercise 4: Mixing Boolean and Fancy Indexing
# arr = np.array([[5, 15, 25],  
#                 [35, 45, 55],  
#                 [65, 75, 85]])
# Tasks:
# 1. Extract all numbers that are multiples of 5 and greater than 30.
# 2. Replace all numbers less than 50 with -1.
# 3. Extract the last column and replace all values with 100.
# 4. Create a new array where all values greater than 60 are doubled.
arr = np.array([[5, 15, 25],  
                [35, 45, 55],  
                [65, 75, 85]])
arr[arr>60]*=2
print(arr)
```

```python
# Given a (5, 5) array (values 1 to 25), 
# use advanced indexing to reorder the rows in the order [4, 2, 0, 3, 1] 
# and the columns in the order [1, 3, 0, 4, 2].
arr1 = np.arange(1, 26, 1)
arr1 = np.reshape(arr1,(5, 5))
print(arr1)
arr2=arr1
arr2 = arr1[[4, 2, 0, 3, 1]]
print(arr2)
arr3=arr2
arr3 = arr3[:,[1, 3, 0, 4, 2]]
print(arr3)
```

```python
# 1. Create a 3D Array Create a 3D array of shape (2, 3, 4) filled with random numbers.
# 2. Reshape above (2,3,4) array into a (3,2,4) array.
# 3. Convert the above 3D array into a 1D array.
# 4. Replace all even values in the second depth with the value -1.
# 5. Check if the array contains atleast a number greater than 95 or not.
# 6. Convert 4th column of first depth with the value 10.

array1 = np.random.randint(1,100,(2, 3, 4))
print(array1, '\n\n===============\n')

array2 = np.reshape(array1, (3,2,4))
print(array2, '\n\n===============\n')

array3 = array1.flatten()
print(array3, '\n\n===============\n')

array4 = array1
array4[1][array4[1]%2==0]=-1
print(array4, '\n\n===============\n')
```
