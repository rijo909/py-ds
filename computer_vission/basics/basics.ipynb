```python
cap=cv2.VideoCapture(1)
ret, frame = cap.read()
if ret:
    cv2.imshow('camera', frame)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    cap.release()
else:
    print('No Frame')
```

```python
cap=cv2.VideoCapture(1)
while True:
    ret, frame = cap.read()
    if not ret:
        print('No Frame')
        break
    else:
        cv2.imshow('camera', frame)
        if cv2.waitKey(200)==ord('a'):
            cv2.destroyAllWindows()
            cap.release()
            break
        
```
