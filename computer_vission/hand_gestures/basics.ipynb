```python
def count_fingers(landmarks):
    tip=[8, 12, 16, 20]
    pip=[6, 10, 14, 18]
    c=0
    for t,p in zip(tip,pip):
        t_x,t_y=landmarks.landmark[t].x, landmarks.landmark[t].y
        p_x,p_y=landmarks.landmark[p].x, landmarks.landmark[p].y
        if t_y<p_y:
            c+=1
    pinky_tx=landmarks.landmark[20].x
    thumb_tx=landmarks.landmark[4].x
    thumb_px=landmarks.landmark[2].x
    # right hand
    if thumb_tx > pinky_tx:
        if thumb_tx>thumb_px:
            c+=1
    else:
        if thumb_tx<thumb_px:
            c+=1
    return c
```

```python
def peace(landmarks):
    c=count_fingers(landmarks)
    index_ty,index_py=landmarks.landmark[8].y,landmarks.landmark[6].y
    middle_ty,middle_py=landmarks.landmark[12].y,landmarks.landmark[10].y
    if c==2 and index_ty<index_py and middle_ty<middle_py:
        return True
    else: 
        return False
```

```python
def draw(landmarks, frame):
    h, w, _ = frame.shape
    i_x, i_y, i_p = landmarks.landmark[8].x, landmarks.landmark[8].y, landmarks.landmark[6].y
    x=round(i_x*w)
    y=round(i_y*h)
    if count_fingers(landmarks)==1 and i_y<i_p:
        return (x,y)
    else:
        return False
```
