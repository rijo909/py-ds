```python
import cv2
```

```python
face_cascade=cv2.CascadeClassifier(r'C:\xampp816\htdocs\personal-dev\py-ds\computer_vission\face_detection\haarcascade_frontalface_default (1).xml')
```

```python
img_array=cv2.imread(r'C:\xampp816\htdocs\personal-dev\py-ds\computer_vission\face_detection\rijo_kurian.jpg')
img_res=cv2.resize(img_array, (600,600))
```

```python
import matplotlib.pyplot as plt
img_res=cv2.cvtColor(img_res, cv2.COLOR_BGR2RGB)
plt.imshow(img_res)
```

```python
res=face_cascade.detectMultiScale(img_res, minNeighbors=5)
if len(res)!=0:
    x1, y1, w, h = res[0]
    x2=x1+w
    y2=y1+h
    cv2.rectangle(img_res, (x1, y1), (x2, y2), (0, 0, 255), 3)
    cv2.imshow('Face', img_res)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print(2)
```

```python
import cv2
face_cascade=cv2.CascadeClassifier(r'C:\xampp816\htdocs\personal-dev\py-ds\computer_vission\face_detection\haarcascade_frontalface_default (1).xml')

cap=cv2.VideoCapture(1)
while True:
    ret, frame=cap.read()
    if not ret:
        print('no frame')
        break
    height, width, _ = frame.shape
    faces=face_cascade.detectMultiScale(frame, minNeighbors=20)
    frame=cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    if len(faces) != 0:
        for i in faces:
            x,y,w,h=i
            cv2.rectangle(frame, (x, y), (x+w, y+h), (23, 233, 122), 2)
        cv2.putText(frame, f'{len(faces)} faces detected', (100, 30), cv2.FONT_HERSHEY_COMPLEX, 1, (255, 23, 123), 1)
    else:
        cv2.putText(frame, f'No faces detected', (100, 30), cv2.FONT_HERSHEY_COMPLEX, 1, (255, 23, 123), 1)
    cv2.imshow('Face', frame)
    if cv2.waitKey(10)==ord('a'):
        cv2.destroyAllWindows()
        cap.release()
        break
```

```python
import cv2
face_cascade=cv2.CascadeClassifier(r'C:\xampp816\htdocs\personal-dev\py-ds\computer_vission\face_detection\haarcascade_frontalface_default (1).xml')

cap=cv2.VideoCapture(0)
while True:
    ret, frame=cap.read()
    if not ret:
        print('no frame')
        break
    height, width, _ = frame.shape
    faces=face_cascade.detectMultiScale(frame, minNeighbors=20)
    if len(faces) != 0:
        for i in faces:
            x,y,w,h=i
            cv2.rectangle(frame, (x, y), (x+w, y+h), (23, 233, 122), 2)
        cv2.putText(frame, f'{len(faces)} faces detected', (100, 30), cv2.FONT_HERSHEY_COMPLEX, 1, (255, 23, 123), 1)
    else:
        cv2.putText(frame, f'No faces detected', (100, 30), cv2.FONT_HERSHEY_COMPLEX, 1, (255, 23, 123), 1)
    frame=cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    frame=cv2.Canny(frame, 200, 200, cv2.THRESH_BINARY)
    cv2.imshow('Face', frame)
    if cv2.waitKey(10)==ord('a'):
        cv2.destroyAllWindows()
        cap.release()
        break
```
