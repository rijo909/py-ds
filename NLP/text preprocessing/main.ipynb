```python
sentence = "The quick brown foxes were jumping over the lazy dogs, aren't they?"
```

Converting to lower case

```python
sentence = sentence.lower()
sentence
```

Punctuation removal

```python
import string
punct=string.punctuation
sentence = "".join(i for i in sentence if i not in punct)
sentence
```

Tokenisation

```python
tokens=sentence.split()
```

Stop word removal - Natural Language toolkit (nltk)

```python
import nltk 
from nltk.corpus import stopwords
stop=stopwords.words('english')
stop
```

```python
tokens_without_stop_word = [i for i in tokens if i not in stop]
tokens_without_stop_word
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmer.stem('amazing')
stemmed_data = [stemmer.stem(i) for i in tokens_without_stop_word]
stemmed_data
```

```python
# create a function that attcdepts sting as argument
# and should return NLP text preprocssed data 

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

def preprocessing(sentence):
    # lower casing
    res = sentence.lower()
    # remove punctuation
    sentence = "".join(i for i in sentence if i not in punct)
    # token spliting
    tokens=sentence.split()
    # remove stop words
    stop=stopwords.words('english')
    tokens_without_stop_word = [i for i in tokens if i not in stop]
    # lematizer
    lemmatizer = WordNetLemmatizer()
    lemmatized_data = [lemmatizer.lemmatize(i, 'v') for i in tokens_without_stop_word]
    res = ' '.join(lemmatized_data)
    return res
sentence_list = [
                    'Daisy teaches English',
                    'Anitha called me yesterday',
                    'The girl wearing the yellow dress is my new neighbour',
                    'hello world, have a nice day!',
                ]
result = []
for i in sentence_list:
    result.append(preprocessing(i))
result
```
