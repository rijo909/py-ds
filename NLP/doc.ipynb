Count Vectorizer
A count vectorizer is a feature extraction technique from
a text document into a vector matrix of token counts (Bank of words)

Text preprocessing


Must be root words

must be in single case as vectorising is case sensitive

should remove special characters like punctuations

Steps :
1. lowercasing
2. strip out punctuation marks
3. Tokenisation - split words into individual words
4. Removing stop words
5. Stemming or Lemmatization - reduce words to their base or root form

Stemming - croping and fast ---> based on rules

Lemmatization - smarter and slow ---> based on dictionary
