```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import string

def preprocessing(sentence):
    return ' '.join([WordNetLemmatizer().lemmatize(w, 'v') for w in "".join([c for c in sentence.lower() if c not in string.punctuation]).split() if w not in stopwords.words('english')])

df['preprocessed_message'] = df['Message'].apply(preprocessing)

df
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
y=le.fit_transform(df['Category'])
```

```python
x=df['preprocessed_message']
x
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer()
x_vector=vectorizer.fit_transform(x)
x_vector.shape
```

```python
x_vector
```

Train Test Split

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x_vector, y)
```

Model training

```python
from sklearn.naive_bayes import MultinomialNB
model=MultinomialNB()
model.fit(x_train, y_train)
y_pred=model.predict(x_test)
```

Evaluation

```python
from sklearn.metrics import classification_report
```

```python
test_text = 'Free free free, grab this offer soon, offer ends soon'
test_pre=preprocessing(test_text)
test = vectorizer.transform([test_text])
res=model.predict(test)
le.inverse_transform([res[0]])[0]
```

Function Creation

```python
def spam_classifier(text):
    text=preprocessing(text)
    vector = vectorizer.transform([text])
    res=model.predict(vector)
    return le.inverse_transform([res[0]])[0]
text = 'Free free free, grab this offer soon, offer ends soon'
text = 'goodmorning all!'
spam_classifier(text)
```
