Ensemble Technique:
Ensemble Technique in machine learning refers to the process of combining predictions from multiple models to improve overall performance, accuracy and robustness compared to individual models,The code idea is thet a group of weak leaners can come together to form a strong learner.

Individual models---> Base model, Weak learner

    Bootstrap sampling is a statistical technique used to create multiple datasets by randomly sampling with replacement from an original dataset. It's the foundation of bagging in ensemble learning.
    Key Features of Bootstrap Sampling:
    Sampling with replacement: Each sample is chosen randomly, and the same item can be selected more than once.
    Size: The bootstrap sample is typically the same size as the original dataset.
    Purpose: Estimate variability, confidence intervals, or create diverse training sets for ensemble models.

gradient decent algorithm
gradient decent algorithm is an optimization algorithm used to minimize a loss function

suppose we have a function (like a loss/cost funtion) that we want to minimize. Gradient decent helps us find th minimum point by taking small steps in the direction where the function decreases the fastest.
1. Initial parameters (weights) randomly.
2. Comute the gradient of the loss funtion with resect to the parameters.
3. Update the parameters
Parameters in gradient decent algorithm :
1. n_estimators: number of boosting stages to perform. default value is 100.
2. learning_rate: smaler values need more trees, but can improve performance.
3. max_depth: Maximum depth of each tree. Controlls complexity; deeper trees can capture more patterns, but may over fit. Typical values are 3 to 8.
4. min_sample_split
5. min_samples_leaf
