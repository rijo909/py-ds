
---

1. Introduction to ANN

An Artificial Neural Network is a computational model inspired by the structure and functioning of the human brain. It consists of interconnected layers of nodes (neurons) that process data in a hierarchical manner.


---

2. Architecture of ANN

Input Layer: Receives the raw data features.

Hidden Layer(s): Performs intermediate computations using weighted connections and activation functions.

Output Layer: Produces the final result (e.g., class label or predicted value).



---

3. Components of ANN

Weights: Parameters that determine the strength of connections between neurons.

Bias: Added to the weighted sum to shift the activation function. To have flexibility in nueral networks.

Activation Function: Introduces non-linearity (e.g., sigmoid, ReLU, tanh).

Loss Function: Measures the error between predicted and actual output.

Optimizer: Adjusts weights and biases to minimize the loss (e.g., SGD, Adam).



---

4. Forward Propagation

Each neuron receives inputs, multiplies by weights, adds bias, and applies the activation function.

Output from one layer becomes input to the next.


Mathematically:

Z = W·X + b  
A = activation(Z)


---

5. Loss Calculation

The difference between predicted output and actual output is calculated using a loss function like:

MSE for regression

Cross-entropy for classification




---

6. Backpropagation

Calculates the gradient of the loss function with respect to each weight using the chain rule.

These gradients indicate how to change the weights to reduce the loss.



---

7. Weight Update

Using an optimization algorithm (like gradient descent), the weights are updated as:


W = W - learning_rate * dL/dW


---

8. Training Process

Repeat forward propagation, loss calculation, backpropagation, and weight update for several epochs.

Use a batch of data at a time (batch or mini-batch training).



---

9. Evaluation

After training, evaluate performance using unseen test data.

Use metrics like accuracy, precision, recall, or RMSE depending on the task.



---

10. Prediction

Use the trained ANN model to make predictions on new data.
