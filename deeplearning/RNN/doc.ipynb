RNN (Recurrent Neural Network)
----

To train sequential data like text data, time series data

Vanilla RNN - has disadvantage of vanishing gradient or vanishing explotion where old information are forgotby the mode if closer to 0

LSTMs (Long Term Short Memory)
They are having 3 gates:

1. Input gate
gate responsible of allowing a sequential data to enter the model
2. Forget gate
Responsible to let what data needs to be forgotten, rather than model deciding it
3. Output gate
responsible for the combinhg of previous output to current data
